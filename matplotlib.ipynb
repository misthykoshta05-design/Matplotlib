```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
price=[48000,54000,57000,49000,47000,45000]
year=[2015,2016,2017,2018,2019,2020]
plt.plot(year,price)
```

```python
batsman=pd.read_csv('sharma-kohli.csv')
plt.plot(batsman['index'],batsman['V Kohli'])
```

```python
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
# TITLE
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
plt.title('Rohit vs Virat Kohli Career Comparison')
plt.xlabel('season')
plt.ylabel('Runs score')
```

```python
# COLOUR AND LINEWIDTH MARKER SIZE
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='dotted',linewidth=3)# solid,dashed
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6',linestyle='dashdot',linewidth=2)
plt.title('Rohit vs Virat Kohli Career Comparison')
plt.xlabel('season')
plt.ylabel('Runs score')
```

```python
#MARKER
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='dotted',linewidth=3,marker='D',markersize=10)
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6',linestyle='dashdot',linewidth=2,marker='+')
plt.title('Rohit vs Virat Kohli Career Comparison')
plt.xlabel('season')
plt.ylabel('Runs score')
```

```python
#LABEL
# plt.legend(loc='best')# for putting label
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='dotted',linewidth=3,marker='D',markersize=10,label='virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6',linestyle='dashdot',linewidth=2,marker='+',label="rohit")
plt.title('Rohit vs Virat Kohli Career Comparison')
plt.xlabel('season')
plt.ylabel('Runs score')
plt.legend()
plt.show()
plt.grid()
```

```python
children=[10,20,40,10,30]
colors=['red','blue','green','yellow','pink']
plt.bar(colors,children,color='black')
```

```python
df=pd.read_csv('batsman_season_record.csv')
```

```python
plt.bar(df['batsman'],df['2015'])
```

```python
plt.bar(df['batsman'],df['2015'],width=0.1)
plt.bar(df['batsman'],df['2016'])
```

```python
plt.bar(np.arange(df.shape[0])-0.5,df['2015'],width=0.75)
```

```python
plt.bar(np.arange(df.shape[0])-0.2,df['2015'],width=0.2,color='yellow')
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2,color='red')
plt.bar(np.arange(df.shape[0])+0.2,df['2017'],width=0.2,color='green')
```

```python
children=[10,20,40,10,30]
colors=['redredredred','blueblueblluue','yekkowyellow','greengreenggfg','pikpikppiknk']
plt.bar(colors,children,color='black')
plt.xticks(rotation='vertical')
```

```python
#STACKED BAR CHART
plt.bar(df['batsman'],df['2017'],label='2017')
plt.bar(df['batsman'],df['2016'],bottom=df['2017'],label='2016')
plt.bar(df['batsman'],df['2015'],bottom=(df['2016']+df['2017']),label='2015')
plt.legend()
plt.show()
```

PIE CHART

```python
data=[23,45,100,49,20]
sub=['eng','hindi','sst','science','sanskrit']
plt.pie(data,labels=sub,autopct='%0.1f',explode=[0.1,0,0,0.2,0],shadow=True)
plt.show()
```
